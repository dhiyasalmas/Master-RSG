# **Program mereplace data input pada Origen**

```python
import pandas as pd
import numpy as np

#Masukkan nilai data yang ingin di rubah
data_variasi = ['10', '20', '30', '40', '50']
#Masukkan baris ke berapa yang ingin di ubah
baris_target = 15
#import file
#data_origen = pd.read_fwf('HTR10.txt')
#Atur baris yang ingin di rubah
for i in range (len(data_variasi)):
    data_origen = pd.read_fwf('HTR.txt')
    baris = list(data_origen.loc[baris_target])
    baris_int = np.char.split(baris[1:]," ").tolist()
    #made to array from 2d to 1d
    baris_int = np.array(baris_int).flatten()
    #Atur mau kolom di baris keberapa yang diubah
    baris_int = np.char.replace(baris_int, baris_int[2], data_variasi[i])
    baris_int = np.insert(baris_int, 0, baris[0])
    print(baris_int)
    #Balikkin jadi 2 nilai array kayak awal
    filtered_array = [x for x in baris_int if x]
    #Gabungin semua
    result = ' '.join(filtered_array)
    #Masukin ke dataframe
    data_origen.iloc[baris_target] = result
    # Export the variation DataFrame to a text file
    file_name = f'htr{data_variasi[i]}.txt'
    data_origen.to_csv(file_name, sep='\t', index=False)  # Adjust separator and index if needed
    print(f"Exported {file_name}")
    i += 1
```
